## <span style="color:blue">3. Slicing of a List in Python
In Python, it is possible to access a section of items from the list using the slicing operator :. For example,

<span style="color:red">Write a Python program to:

- <span style="color:red">Print the items from index 2 to index 4 in my_list.
- <span style="color:red">Print the items from index 5 to the end of my_list.
- <span style="color:red">Print all items in my_list from the beginning to the end.

```python
my_list = ['p', 'r', 'o', 'g', 'r', 'a', 'm']

# items from index 2 to index 4
print(my_list[2:5])

# items from index 5 to end
print(my_list[5:])

# items beginning to end
print(my_list[:])
```

#### <span style="color:red"> Write a Python program using the turtle module and a dictionary that stores fruits and their colors. Display the names of three fruits — Apple, Banana, and Grapes — on the screen in their respective colors (red, yellow, purple) at different positions.

```python

import turtle

fruits = {
    "apple": "red",
    "banana": "yellow",
    "grapes": "purple"
}

t = turtle.Turtle()
t.hideturtle()
t.penup()

t.color(fruits["apple"])
t.goto(-100, 50)
t.write("Apple", font=("Arial", 30, "bold"))

t.color(fruits["banana"])
t.goto(0, 0)
t.write("Banana", font=("Arial", 30, "bold"))

t.color(fruits["grapes"])
t.goto(100, -50)
t.write("Grapes", font=("Arial", 30, "bold"))

turtle.done()
```
